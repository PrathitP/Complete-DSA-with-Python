# Variables 

- Helps to store value in runtime

```python
## Declaring and Assigning Variables
age = 23
name = "Prathit"
height = 6.1
is_student = True


## Printing the variables 

print("Age :",age)
print("Name :",name)
print("Height :",height)
```

```python
## Naming Conventions
## Variable Names should be descriptive
## They must always start with a letter or an underscore and can contains letter,numbers and underscores
## Variable names are case sensitive


# Valid Variable Name

first_name = "Prathit"
last_name = "Panda"
```

```python
print(first_name + last_name)
```

```python
## Invalid Variable Names

# 2age = 45
# first-name = "Prathit"
@Name = "Prathit"
```

```python
## Case Sensitivity 

name = "Prathit"
Name = "Panda"
```

```python
## Understanding Variables types
## Python is dynamically typed , type of a variable is determined at runtime 
 
age = 23 # int 
name = "Prathit" # string
height = 6.1 # float
is_student = True # bool

print(type(name))
print(type(age))
print(type(height))
```

### Type Conversion

```python
age = 25
print(type(age))

## Type Conversion

age_str = str(age)
print(age_str)
print(type(age_str))
```

```python
age = '24'
print(type(age))
print(type(int(age)))
```

```python
name = "Prathit"
name_int = int(name)
print(type(name_int))
```

```python
height = 6.23
height_int = int(height)
print(type(height_int))
```

```python
age = 27
age_float = float(age)
print(type(age_float))
print(age_float)
```

```python
## Dynamic Typing
# Python allows the type of a variable to change as the program executes

var = 10 
print(var,type(var))

var = "Hello"
print(var,type(var))

var = 3.14
print(var,type(var))
```

```python
string_number = "10"
print(string_number)
print(type(string_number))
integer_number = int(string_number)
print(integer_number)
print(type(integer_number))
```

### Input

```python
#Taking the Input 

age = input("What is your age")
print("The age is ",age)
print(type(age))
```

```python
# How to take a particular type of variable input 
age = int(input("What is your age"))
print("The age is ",age)
print(type(age))
```
