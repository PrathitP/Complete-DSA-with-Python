## Arthemetic Operators

```python
## Arthemtic Operations 
a = 10
b = 5

add_result = a+b # Addtion
sub_result= a-b # Subtraction
mult_result = a*b # Multiplication
div_result = a/b # Division
floor_div = a//b # Floor Division # It will remove the decimal and give the floor value which is close to the round off to exact decimal before point
modulus_result = a%b # Modulus Division
exponential_result = a**b # Exponents

print("Addition :",add_result)
print("Subtraction :",sub_result)
print("Multiplication :",mult_result)
print("Division :",div_result)
print("Floor Division :",floor_div)
print("Modulus :",modulus_result)
print("Exponential :",exponential_result)
```
