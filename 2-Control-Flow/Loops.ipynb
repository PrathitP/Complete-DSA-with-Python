- While Loop 

```python
## While loop 
# It continues to execute until the conditon is true 


count = 0

while count < 5 :
    print(count)
    count = count + 1 
```

```python
count = 0 

while count%2 == 0 :
    print(count)
    count = count + 1
```

### Break 

```python
### Loop Control Statement 

### Break 

### The break statement exist the loop prematurely 

for i in range(10):
    print(i)
    if i == 5 :
        break 
```

```python
### Continue 


### Skips the current iteration and continues with the rest 

## its like skipping the current loop for 1 time and executing the loop from next one 


for i in range(10) :
    if i%2== 0 :
        continue
    print(i)
```

### Pass

```python
### Pass 
### The pass is a null operation ;
### In short it does nothing just to have code intact like if we have empty function
for i in range(10):
    if i%2==0:
        print("The numer is ",i)
        pass 
    print(i)
```

### Nested Loops 

```python
### Nested Loops 

## The loop inside  a loop 


for i in range(5):
    for j in range (4): ### First it will complete the inner loop and then increment the outer loop 
        print(f"i:{i},j:{j}")
```

```python
## Calcluate the sum of first N=10  natural Numbers 

### While Loop 

number = 10 
sum = 0 
count = 0 
while count < number + 1 :
    sum = sum + count 
    count = count + 1 
print(f"The sum of first {number} natural numbers is sum :{sum}")
```

# Important and Interesting Qs

```python
### Example - Prime numbers between 1 and 100 
## With for loop we can always run a else block 


for num in range(101):
    if num>1 :
        for i in range(2,num):
            if num%i==0:
                break
        else :
            print(num)
```
