# Conditional Statement

```python
### If Statment

age = 20 

if age>=18 :
    print("You are allowed to apply for Driving License")
```

```python
## Else 
age = 15 

if age>=18 :
    print("You are allowed to apply for Driving License")
else :
    print("You need to be atleast 18 years of Age to Apply. !!")
```

```python
### elif Statement 
## It stands for else-if 

age = 17

if age <18 :
    print("You are a Kiddo !")
elif age <19 :
    print("You can Apply for DL")
else :
    print("You can apply happily ")
```

```python
## Nested Conditional Statements 

# Number to check if it even odd or negative

number = int(input("Enter the Number here :"))

if number > 0 :
    print("The number is Positive Number")
    if number%2 == 0:
        print("Even Number")
    else :
        print("Odd Number")
elif number == 0 :
    print("The Number is Zero an integer neither positive or negative")
else :
    print("It is a Negative Number")
```

```python
## Practical Examples

## Determine is a leap year (Any year which that is divisible by 4 is a leap year)

year = int(input("Enter a year :"))

if year%4==0 and year%100!=0 :
    print("Leap Year..!!")
elif year%400==0 :
    print("Leap Year..!!")
else :
    print("Not a Leap Year")
```

### Other Approach

```python
year = 2026
if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else :
            print(year,"is not a leap year")
    else :
        print(year,"is not a leap year")
else :
    print(year,"is not a leap year")
```

- Assignment : Calculator 

```python
number_1 = int(input("Enter your first operand :"))
number_2 = int(input("Enter your second operand :"))
operation = input("Enter the Operation which you want to perform (+,-,*,/) :")

if operation == '+':
    print("Addition of ",number_1,"and",number_2)
    result = number_1 + number_2
    print("Addition result is :",result)
elif operation == '-':
    print("Substraction of ",number_1,"and",number_2)
    result = number_1 - number_2
    print("Substraction result is :",result)
elif operation == '*':
    print("Multiplication of ",number_1,"and",number_2)
    result = number_1 * number_2
    print("Multliplication result is :",result)
elif operation == '/':
    print("Division of ",number_1,"and",number_2)
    result = number_1 / number_2
    print("Division result is :",result)
else :
    print("Please enter a valid Operation..!! THANK YOU...")
```
